```python
#@title 5. Watch In-Notebook & Auto-Download MP4 to Your Computer
from IPython.display import HTML
from base64 import b64encode
from google.colab import files
import os

target_file = output_path if 'output_path' in locals() and os.path.exists(output_path) else "out/educational-reel.mp4"
if os.path.exists(target_file):
    mp4 = open(target_file, "rb").read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
    display(HTML(f"""
    <div style="display:flex; flex-direction:column; align-items:center; margin-top:10px;">
        <h3 style="color:#00F0FF; font-family:sans-serif;">🎬 Preview (1080x1920 9:16)</h3>
        <video width="300" height="533" controls autoplay style="border-radius:16px; box-shadow:0 10px 30px rgba(0,240,255,0.3); border:2px solid #00F0FF;">
            <source src="{data_url}" type="video/mp4">
        </video>
    </div>
    """))
    print(f"📥 Downloading {target_file} to your computer...")
    files.download(target_file)
else:
    print(f"❌ Output file {target_file} not found. Check step 4 logs.")
```
